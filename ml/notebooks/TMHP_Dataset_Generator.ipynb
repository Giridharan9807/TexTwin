```python
import pandas as pd
import numpy as np
import random

random.seed(42)
np.random.seed(42)

data = []

# Running - 2000
for i in range(2000):
    data.append([
        f"LOOM{i+1:04d}",
        round(random.uniform(32,38),2),
        round(random.uniform(0.5,2.0),2),
        random.randint(1180,1220),
        round(random.uniform(58,62),2),
        round(random.uniform(4.2,5.2),2),
        random.randint(50,500),
        "Running"
    ])

# Maintenance - 1500
for i in range(2000,3500):
    data.append([
        f"LOOM{i+1:04d}",
        round(random.uniform(39,44),2),
        round(random.uniform(2.0,5.0),2),
        random.randint(1150,1185),
        round(random.uniform(55,58),2),
        round(random.uniform(5.2,6.0),2),
        random.randint(500,1200),
        "Maintenance"
    ])

# Critical - 1000
for i in range(3500,4500):
    data.append([
        f"LOOM{i+1:04d}",
        round(random.uniform(45,50),2),
        round(random.uniform(5.0,8.0),2),
        random.randint(1080,1145),
        round(random.uniform(50,55),2),
        round(random.uniform(6.0,7.0),2),
        random.randint(1200,2500),
        "Critical"
    ])

# Stopped - 500
for i in range(4500,5000):
    data.append([
        f"LOOM{i+1:04d}",
        round(random.uniform(28,35),2),
        0,
        0,
        round(random.uniform(50,60),2),
        0,
        random.randint(0,2500),
        "Stopped"
    ])

df = pd.DataFrame(data, columns=[
    "Machine_ID",
    "Temperature",
    "Vibration",
    "RPM",
    "Humidity",
    "Power",
    "Running_Hours",
    "Status"
])

df.head()
```
